```python
# Download a free spoken audio sample to test with
!wget -q "https://www.soundhelix.com/examples/mp3/SoundHelix-Song-1.mp3" -O sample_song.mp3

# Convert mp3 to wav (torchaudio works best with wav)
!pip install -q pydub
!apt-get install -q ffmpeg

from pydub import AudioSegment
audio = AudioSegment.from_mp3("sample_song.mp3")
audio = audio[:10000]  # take first 10 seconds only
audio.export("sample.wav", format="wav")
print("Audio file ready!")
```

```python
import torchaudio
import matplotlib.pyplot as plt
import torch

# Load the audio file
waveform, sample_rate = torchaudio.load("sample.wav")

print("Shape:", waveform.shape)        # (channels, samples)
print("Sample rate:", sample_rate)     # how many samples per second
print("Duration:", waveform.shape[1] / sample_rate, "seconds")

# Plot the waveform
plt.figure(figsize=(12, 3))
plt.plot(waveform[0].numpy())          # [0] = first channel (left ear)
plt.title("Waveform — what audio looks like as numbers")
plt.xlabel("Sample index")
plt.ylabel("Amplitude (-1 to 1)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Whisper needs 16000 Hz sample rate
TARGET_SR = 16000

resampler = torchaudio.transforms.Resample(
    orig_freq=sample_rate,
    new_freq=TARGET_SR
)

waveform_16k = resampler(waveform)

# Convert stereo (2 channels) to mono (1 channel) by averaging
if waveform_16k.shape[0] > 1:
    waveform_16k = waveform_16k.mean(dim=0, keepdim=True)

print("New shape:", waveform_16k.shape)
print("New sample rate: 16000 Hz")
print("Duration:", waveform_16k.shape[1] / TARGET_SR, "seconds")
```

```python
# Compute mel spectrogram — this is Whisper's "vision" of audio
mel_transform = torchaudio.transforms.MelSpectrogram(
    sample_rate=TARGET_SR,
    n_fft=400,
    hop_length=160,
    n_mels=80           # Whisper uses 80 mel frequency bins
)

mel_spec = mel_transform(waveform_16k)

# Convert to log scale (matches how humans perceive loudness)
log_mel = torchaudio.functional.amplitude_to_DB(
    mel_spec, multiplier=10, amin=1e-10, db_multiplier=0
)

print("Mel spectrogram shape:", log_mel.shape)  # (1, 80, time_frames)

# Plot it
plt.figure(figsize=(12, 4))
plt.imshow(
    log_mel[0].numpy(),
    aspect="auto",
    origin="lower",
    cmap="inferno"
)
plt.colorbar(label="dB")
plt.title("Mel Spectrogram — what Whisper sees (bright = loud frequency)")
plt.xlabel("Time frames")
plt.ylabel("Mel frequency bins (low → high pitch)")
plt.tight_layout()
plt.show()
```

```python
from transformers import WhisperProcessor, WhisperForConditionalGeneration

print("Loading Whisper Small... (downloads ~460MB first time)")
processor = WhisperProcessor.from_pretrained("openai/whisper-small")
model = WhisperForConditionalGeneration.from_pretrained("openai/whisper-small")
model = model.to("cuda")
model.eval()
print("Whisper loaded! Parameters:", round(sum(p.numel() for p in model.parameters()) / 1e6), "M")
```

```python
# Prepare audio for Whisper
audio_input = waveform_16k.squeeze().numpy()  # remove channel dim, convert to numpy

# Whisper's processor converts raw audio → input features
inputs = processor(
    audio_input,
    sampling_rate=TARGET_SR,
    return_tensors="pt"   # pt = PyTorch tensors
)

input_features = inputs.input_features.to("cuda")

# Generate transcription
print("Transcribing...")
with torch.no_grad():    # no_grad = don't compute gradients (saves memory)
    predicted_ids = model.generate(input_features)

# Decode token IDs back into text
transcription = processor.batch_decode(
    predicted_ids,
    skip_special_tokens=True
)[0]

print("\n--- Whisper Output ---")
print(transcription)
```

```python
from jiwer import wer, cer

# For this music sample we don't have ground truth lyrics
# so let's test WER with a made-up example to understand the metric
reference  = "i love to sing songs every day"   # ground truth
hypothesis = "i love to sing song every day"    # what model said (missing 's')

word_error_rate = wer(reference, hypothesis)
char_error_rate  = cer(reference, hypothesis)

print(f"WER:  {word_error_rate:.1%}")   # e.g. 14.3% — 1 word wrong out of 7
print(f"CER:  {char_error_rate:.1%}")

print("\nWER = (substitutions + deletions + insertions) / total words")
print("Lower is better. 0% = perfect. 100% = completely wrong.")
```
